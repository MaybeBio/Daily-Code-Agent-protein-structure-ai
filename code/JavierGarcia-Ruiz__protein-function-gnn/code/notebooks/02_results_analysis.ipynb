# 02: Model Benchmark & Representation Analysis

This notebook benchmarks **1D Dilated CNN (sequence baseline)** against **Contact GCN** and **Edge-Attentive GAT (3D graph models)** on held-out test proteins (<30% sequence homology).


```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from protgnn.data.constants import EC_CLASSES, EC_DESCRIPTIONS
from protgnn.visualization.plotting import set_style

set_style()
bench_path = Path('../reports/results/metrics_benchmarks.csv')
df_bench = pd.read_csv(bench_path)
df_bench
```

## 1. Multi-Metric Benchmark Comparison


```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4.5))
metrics = [('macro_f1', 'Macro-F1'), ('macro_mcc', 'Macro-MCC'), ('macro_aupr', 'Macro-AUPR')]
for ax, (m, label) in zip(axes, metrics):
    sns.barplot(data=df_bench, x='experiment', y=m, ax=ax, palette='Blues_d')
    ax.set_title(label)
    ax.set_ylim(0, 1.0)
plt.tight_layout()
plt.show()
```
