### Data proccessing function

```python
def get_data(data_name):
    with open(data_name,'r') as data:
        return data.read()
```

```python
# 1. Dictionnaires de Mapping
# Acides aminés standards
AA_MAP = {
    'A': 0, 'C': 1, 'D': 2, 'E': 3, 'F': 4, 'G': 5, 'H': 6, 'I': 7, 'K': 8, 'L': 9, 
    'M': 10, 'N': 11, 'P': 12, 'Q': 13, 'R': 14, 'S': 15, 'T': 16, 'V': 17, 'W': 18, 'Y': 19,
    '<PAD>': 20  # Le Spacer pour les bords
}

# Labels: h=Helix, e=Sheet, _=Coil
LABEL_MAP = {'h': 0, 'e': 1, '_': 2}
```

```python


def parse_raw_data(raw_text):
    """
    Lit le texte brut et retourne une liste de protéines.
    Chaque protéine est un tuple (liste_indices_aa, liste_indices_labels)
    """
    proteins = []
    current_aa_seq = []
    current_label_seq = []
    
    lines = raw_text.strip().split('\n')
    
    for line in lines:
        line = line.strip()
        if not line: continue
        
        if line == '<>':
            # Nouvelle protéine : sauvegarde la précédente si elle existe, puis réinitialise
            if current_aa_seq:
                proteins.append((current_aa_seq, current_label_seq))
            current_aa_seq = []
            current_label_seq = []
        else:
            # Ligne de donnée : "E _" ou "F h"
            parts = line.split()
            if len(parts) == 2:
                aa, label = parts[0], parts[1]
                # Conversion immédiate en entiers
                if aa in AA_MAP:
                    current_aa_seq.append(AA_MAP[aa])
                    current_label_seq.append(LABEL_MAP[label])
    
    # Sauvegarde la dernière protéine si elle existe
    if current_aa_seq:
        proteins.append((current_aa_seq, current_label_seq))
    
    return proteins

```

```python
def create_windows(proteins, window_size=13):
    """
    Transforme les séquences complètes en fenêtres glissantes.
    Retourne X (inputs) et Y (targets).
    """
    half_window = window_size // 2
    X = []
    Y = []
    
    for aa_seq, label_seq in proteins:
        # Longueur de la protéine
        L = len(aa_seq)
        
        # On ajoute du Padding (Spacer) au début et à la fin de la séquence
        # pour pouvoir centrer la fenêtre sur le premier et le dernier élément
        padded_seq = [AA_MAP['<PAD>']] * half_window + aa_seq + [AA_MAP['<PAD>']] * half_window
        
        # Glissement de la fenêtre
        for i in range(L):
            # i correspond à la position du résidu central dans la séquence originale
            # Dans padded_seq, le résidu central est à l'index i + half_window
            window = padded_seq[i : i + window_size]
            target = label_seq[i]
            
            X.append(window)
            Y.append(target)
            
    return torch.tensor(X, dtype=torch.long), torch.tensor(Y, dtype=torch.long)

```

```python
import torch
import torch.nn as nn

class ProteinBiLSTM(nn.Module):
    def __init__(self, vocab_size=21, embedding_dim=42, hidden_dim=64, num_layers=2):
        super().__init__()
        
        # 1. EMBEDDING
        # Transforme les indices (0-20) en vecteurs denses
        self.embedding = nn.Embedding(vocab_size, embedding_dim)
        
        # batch_first=True : pour avoir [Batch, Seq_Len, Features]
        # bidirectional=True : Le modèle devient 2x plus large (64 * 2 = 128 sorties)
        self.lstm = nn.LSTM(
            input_size=embedding_dim,
            hidden_size=hidden_dim,
            num_layers=num_layers,
            batch_first=True,
            bidirectional=True,
            dropout=0.3 # Dropout interne au LSTM
        )
        
        # L'entrée est hidden_dim * 2 car on a concaténé Forward + Backward
        self.fc = nn.Sequential(
            nn.Linear(hidden_dim * 2, 128),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(128,40),
            nn.ReLU(),
            nn.Dropout(0.6),
            nn.Linear(40,3), # 3 classes : Hélice, Feuillet, Autre
        )

    def forward(self, x):
        # x shape : [Batch, Window_Size] (ex: [64, 13])
        
        # 1. Embedding
        x = self.embedding(x) # -> [Batch, 13, 42]
        
        # 2. Passage LSTM
        # out contient les états cachés pour chaque pas de temps
        # out shape : [Batch, 13, 128] (128 car 64*2 directions)
        out, (hn, cn) = self.lstm(x)
        
        # 3. Stratégie de Pooling
        # On veut prédire la structure de l'acide aminé CENTRAL de la fenêtre.
        # Dans une fenêtre de 13, c'est l'index 6 (le 7ème élément).
        # On récupère donc uniquement l'info de ce moment-là.
        
        mid_index = x.shape[1] // 2  # Index 6 pour window 13
        central_features = out[:, mid_index, :] # -> [Batch, 128]
        
        # 4. Classification
        return self.fc(central_features)
```
